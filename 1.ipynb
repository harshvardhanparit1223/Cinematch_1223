```python
import nltk
from nltk.corpus import stopwords
from nltk.stem import WordNetLemmatizer
import re

nltk.download("stopwords")

stop_words = set(stopwords.words("english"))

lemmatizer = WordNetLemmatizer()
```

```python
def process(text):
    # Convert to lowercase
    text = str(text).lower()

    # Remove punctuation and numbers
    text = re.sub(r'[^a-zA-Z\s]', '', text)

    # Split into words
    words = text.split()

    # Remove stopwords
    words = [word for word in words if word not in stop_words]

    # Lemmatize words
    words = [lemmatizer.lemmatize(word) for word in words]

    return " ".join(words)
```

```python
df["tags"] = df["tags"].apply(process)
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
```

```python
tfidf = TfidfVectorizer(max_features=50000,ngram_range=(1,2),stop_words="english")
```

```python
tfidf_matrix =tfidf.fit_transform(df["tags"])
```

```python
tfidf_matrix
```

```python
from sklearn.metrics.pairwise import cosine_similarity
```

```python
def recommended(title, n=10):
    if title not in indices:
        return "Movie not found"

    idx = indices[title]

    # Calculate similarity of the selected movie with all movies
    sim_score = cosine_similarity(tfidf_matrix[idx], tfidf_matrix).flatten()

    # Get indices of most similar movies
    sim_idx = sim_score.argsort()[::-1][1:n+1]

    # Return movie titles
    return df["title"].iloc[sim_idx]
```

```python
recommended("Avatar")
```
